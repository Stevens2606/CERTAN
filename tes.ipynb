```python
import numpy as np

# Array dasar
a = np.array([1, 2, 3, 4])

# Membuat matriks
a = np.array([(1, 2, 3), (4, 5, 6)])

# Matriks khusus
np.zeros([5, 5])
np.identity(3)
np.ones([5, 5])

# Operasi elementwise
a = np.array([1, 2, 3, 4, 5])
b = np.array([6, 7, 8, 9, 10])
print(a + b, a - b, a * b, a / b, a ** 2)

# Perkalian matriks (bukan elementwise)
hasil = np.dot(a, b)

# Indexing & slicing
a = np.arange(10) ** 3
print(a[0], a[-1], a[0:4])

# Transpose, flatten, stacking
a = np.array([(1, 2, 3), (4, 5, 6)])
a.T
a.ravel()
np.hstack((a, a))
np.vstack((a, a))

# Statistik
a = np.floor(np.random.randn(8, 7) * 10)
print(a.max(), a.argmax(), a.min(), a.argmin())
np.sort(a)

# Aljabar linear
a = np.array([(1, -1), (1, 1)])
np.linalg.inv(a)
np.linalg.det(a)

A = np.array([(1, -6), (2, -5)])
Y = np.array([-35, 21])
X = np.linalg.solve(A, Y)
```
