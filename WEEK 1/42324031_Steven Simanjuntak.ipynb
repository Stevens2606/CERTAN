```python
#solusi persamaan linear
A = np.array([[1, -6], [2, -5]])
Y = np.array([-35, 21])
print(A)
print(Y)

A_inv = np.linalg.inv(A)

#menyelesaikan persamaan linear
X1 = np.dot(A_inv, Y)
print(X1)

#cara lain
X2 = np.linalg.solve(A, Y)
print(X2)
```

```python
def f(x,y):
    return np.cos(np.exp(-x**2-y**2))*np.sin(np.sin(np.sqrt(x**2+y**2)))

x = np.linspace(-np.pi, np.pi, 30)
y = np.linspace(-np.pi, np.pi, 30)

X,Y = np.meshgrid(x,y)
Z = f(X,Y)

fig = plt.figure()
ax = plt.axes(projection='3d')
ax.contour3D(X, Y, Z, 50, cmap = 'binary')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z');
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Membuat range x dan y
x = np.linspace(-2, 2, 400)
y = np.linspace(-2, 2, 400)

X, Y = np.meshgrid(x, y)

# Persamaan
Z = (X**6 
     + 3*X**4*Y**2 
     - 3*X**4 
     + 3*X**2*Y**4 
     - X**2*Y**3 
     - 6*X**2*Y**2 
     + 3*X**2 
     + Y**6 
     - 3*Y**4 
     + 3*Y**2)

# Membuat grafik 2 dimensi
plt.contour(X, Y, Z, levels=[1])

plt.xlabel('x')
plt.ylabel('y')
plt.title('Grafik 2 Dimensi')
plt.grid()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

from mpl_toolkits import mplot3d

# Membuat range x dan y
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)

X, Y = np.meshgrid(x, y)

# Persamaan
Z = (5*Y/4 - np.abs(X))**2 + X**2

# Membuat grafik 3 dimensi
fig = plt.figure()

ax = plt.axes(projection='3d')

ax.plot_surface(X, Y, Z)

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')

ax.set_title('Grafik 3 Dimensi')

plt.show()
```
