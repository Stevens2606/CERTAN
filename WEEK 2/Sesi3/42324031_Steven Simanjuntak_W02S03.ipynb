```python
def remove_outliers_sequentially(df2, columns):
    df_cleaned = df2.copy()
    for col in columns:
        q1, q3 = np.percentile(df_cleaned[col], [25, 75])
        iqr = q3 - q1
        lower_bound = q1 - (1.5 * iqr)
        upper_bound = q3 + (1.5 * iqr)
        df_cleaned = df_cleaned[(df_cleaned[col] >= lower_bound) & (df_cleaned[col] <= upper_bound)]
    return df_cleaned

# Terapkan fungsi untuk menghapus outliers dari "Height" dan "Weight"
clean_data = remove_outliers_sequentially(df2, ['Height', 'Weight'])

# Plot boxplots setelah menghapus outliers
columns_to_plot = ['Weight', 'Height']
fig, axs = plt.subplots(len(columns_to_plot), 1, dpi=95, figsize=(7, 5 * len(columns_to_plot)))

if len(columns_to_plot) == 1:
    axs = [axs]
for i, col in enumerate(columns_to_plot):
    axs[i].boxplot(clean_data[col], vert=False)
    axs[i].set_ylabel(col)

plt.show()
```

```python
import pandas as pd
titanic = pd.read_csv("Titanic-Dataset.csv")
titanic.isnull().sum()
```

### Cek Missing Value di Dataset Titanic

Pas dicek pakai isnull().sum(), ternyata ada 3 kolom yang datanya bolong-bolong. Kolom 
Age ada 177 yang kosong, Cabin parah banget sampai 687 baris kosong (hampir 77% dari 
total data), terus Embarked cuma 2 doang yang kosong.

Karena Cabin kosongnya udah kebangetan banyak, kayaknya percuma juga kalau dipaksa diisi, 
soalnya datanya udah nggak representatif lagi. Jadi mending kolom ini langsung dibuang 
aja daripada dipertahanin.

```python
titanic['Age'] = titanic['Age'].fillna(titanic['Age'].median())
titanic['Embarked'] = titanic['Embarked'].fillna(titanic['Embarked'].mode()[0])
titanic = titanic.drop(columns=['Cabin'])
titanic.isnull().sum()
```

### Hasil Setelah Handling

Setelah Age diisi pakai median, Embarked diisi pakai modus, dan Cabin didrop, sekarang 
semua kolom sudah 0 missing value. Age dan Embarked masih dipertahankan karena data yang 
hilang cuma sedikit, jadi masih worth diisi. Cabin didrop karena kalau dipaksa diisi 
malah bisa bikin data jadi nggak akurat, soalnya yang kosong kebanyakan.

```python
diabetes = pd.read_csv("diabetes.csv")
diabetes[['Glucose', 'BloodPressure', 'Insulin']].describe()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, axs = plt.subplots(1, 3, figsize=(15, 5))
sns.boxplot(y=diabetes['Glucose'], ax=axs[0])
sns.boxplot(y=diabetes['BloodPressure'], ax=axs[1])
sns.boxplot(y=diabetes['Insulin'], ax=axs[2])
plt.tight_layout()
plt.show()
```

### Deteksi Outlier: Glucose, BloodPressure, Insulin

Dari boxplot di atas keliatan kalau Insulin punya outlier paling banyak dan paling 
ekstrem, banyak titik yang jauh di atas whisker atas. BloodPressure juga ada beberapa 
outlier, termasuk yang nilainya 0 (kemungkinan data error, soalnya nggak mungkin 
tekanan darah orang 0). Glucose relatif lebih rapi, outlier-nya nggak sebanyak dua 
kolom lainnya.

```python
def remove_outliers(df, columns):
    df_clean = df.copy()
    for col in columns:
        q1, q3 = df_clean[col].quantile([0.25, 0.75])
        iqr = q3 - q1
        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr
        df_clean = df_clean[(df_clean[col] >= lower) & (df_clean[col] <= upper)]
    return df_clean

diabetes_clean = remove_outliers(diabetes, ['Glucose', 'BloodPressure', 'Insulin'])
print("Jumlah data sebelum:", diabetes.shape[0])
print("Jumlah data sesudah:", diabetes_clean.shape[0])
```

```python
fig, axs = plt.subplots(1, 3, figsize=(15, 5))
sns.boxplot(y=diabetes_clean['Glucose'], ax=axs[0])
sns.boxplot(y=diabetes_clean['BloodPressure'], ax=axs[1])
sns.boxplot(y=diabetes_clean['Insulin'], ax=axs[2])
plt.tight_layout()
plt.show()
```

### Kesimpulan

Setelah outlier-nya dibersihin pakai metode IQR, jumlah datanya berkurang dari yang 
awalnya 768 baris jadi tinggal sekitar 700-an baris aja (angka pastinya bisa dicek 
dari hasil print di atas). Kalau dilihat dari boxplot yang baru, titik-titik outlier 
di ketiga kolom udah berkurang jauh, apalagi di Insulin yang tadinya paling parah 
banyak outlier-nya. Jadi bisa dibilang data sekarang udah lebih bersih dan siap 
dipakai buat analisis atau modeling selanjutnya.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.ensemble import RandomForestClassifier
import numpy as np
from mpl_toolkits.mplot3d import Axes3D
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, roc_curve, auc, roc_auc_score
from sklearn.preprocessing import label_binarize
from sklearn.multiclass import OneVsRestClassifier

# Load dataset
df = sns.load_dataset("iris")

# 1. Histogram → Distribusi
plt.figure(figsize=(6,4))
sns.histplot(df['petal_length'], bins=20, kde=True)
plt.title("Histogram: Distribution of Petal Length")
plt.show()
```

```python
# 8. 3D Scatter Plot → 3 features visualization
fig = plt.figure(figsize=(7,5))
ax = fig.add_subplot(111, projection="3d")
species = df["species"].unique()
colors = ["r", "g", "b"]
for sp, col in zip(species, colors):
 subset = df[df["species"] == sp]
 ax.scatter(subset["sepal_length"], subset["sepal_width"], subset["petal_length"],
          c=col, label=sp, s=50)
ax.set_xlabel("Sepal Length")
ax.set_ylabel("Sepal Width")
ax.set_zlabel("Petal Length")
ax.set_title("3D Scatter Plot: Sepal Length, Sepal Width, Petal Length")
ax.legend()
plt.show()
```

```python
# 9. Model-Specific Visualization (e.g Modeling using RandomForest)
# Split data
X = df.drop("species", axis=1)
y = df["species"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
# Train model
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
# 9.1. Confusion Matrix
cm = confusion_matrix(y_test, y_pred, labels=model.classes_)
plt.figure(figsize=(6,4))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=model.classes_, yticklabels=model.classes_)
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
from sklearn.preprocessing import label_binarize
from sklearn.multiclass import OneVsRestClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt

# 9.2. ROC–AUC Curve (multi-class with One-vs-Rest)
# Binarize labels
y_bin = label_binarize(y, classes=model.classes_)
n_classes = y_bin.shape[1]

# Train OvR classifier
ovr_model = OneVsRestClassifier(RandomForestClassifier(n_estimators=100, random_state=42))
ovr_model.fit(X_train, label_binarize(y_train, classes=model.classes_))
y_score = ovr_model.predict_proba(X_test)
y_test_bin = label_binarize(y_test, classes=model.classes_)

# Plot ROC curve
plt.figure(figsize=(7,5))
for i in range(n_classes):
    fpr, tpr, _ = roc_curve(y_test_bin[:, i], y_score[:, i])
    roc_auc = auc(fpr, tpr)
    plt.plot(fpr, tpr, label=f"Class {model.classes_[i]} (AUC = {roc_auc:.2f})")

plt.plot([0,1], [0,1], "k--")  # diagonal
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC–AUC Curve (One-vs-Rest)")
plt.legend()
plt.show()
```
