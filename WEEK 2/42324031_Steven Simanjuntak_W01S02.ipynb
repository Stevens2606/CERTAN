### Analisa Data Previous Purchases


Setelah dihitung, rata-rata pembelian sebelumnya dari pelanggan itu sekitar 25 kali (25.35 
tepatnya), tapi standar deviasinya cukup tinggi yaitu 14.45. Ini artinya jumlah pembelian 
tiap pelanggan sebenarnya beda-beda jauh, ada yang jarang belanja ada juga yang sering 
banget. Kalau dilihat dari nilai minimumnya, ada pelanggan yang baru belanja 1 kali, 
sedangkan yang paling banyak sampai 50 kali.

Untuk kuartilnya, 25% pelanggan (Q1) punya riwayat pembelian di bawah 13 kali, separuh 
data (Q2/median) ada di angka 25, dan 75% pelanggan (Q3) berada di bawah 38 kali pembelian. 
Karena median-nya (25) hampir sama persis dengan mean (25.35), berarti data ini lumayan 
seimbang, tidak terlalu miring ke kiri atau ke kanan.

```python
df[['Age', 'Purchase Amount (USD)']].corr()
```

### Korelasi Age vs Purchase Amount

Angka korelasinya cuma -0.010424, itu udah deket banget sama 0. Artinya umur pelanggan 
nggak ada kaitannya sama besar kecilnya uang yang mereka belanjakan. Yang tua atau yang 
muda, pola belanjanya nggak beda jauh kalau dilihat dari sisi nominal.

```python
df[['Purchase Amount (USD)', 'Review Rating']].corr()
```

### Korelasi Purchase Amount vs Review Rating

Sama kayak yang tadi, korelasi Purchase Amount (USD) sama Review Rating juga kecil 
banget, cuma 0.03. Jadi besar kecilnya uang yang dikeluarin pelanggan nggak nentuin 
mereka bakal kasih rating bagus atau jelek. Yang belanja mahal belum tentu puas, 
yang belanja murah juga belum tentu kecewa.

### Kesimpulan

Dari dua hasil korelasi di atas, keliatan jelas kalau Age dan Purchase Amount 
sama-sama nggak punya pengaruh berarti ke variabel lainnya. Kemungkinan ada faktor 
lain yang lebih nentuin perilaku belanja dan kepuasan pelanggan di dataset ini, 
bukan cuma dari umur atau berapa uang yang mereka habiskan.
